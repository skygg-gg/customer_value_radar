## Construction du dataset consolidé

On a chevauchement de 22 523 lignes entre les deux feuilles pour la période du 1er au 9 décembre 2010.

Afin d'éviter tout doublons, une seule occurrence de cette période est conservée.

Le dataset consolidé contient :
- la feuille 2009-2010 jusqu'au 30 novembre 2010 ;
- l'intégralité de la feuille 2010-2011.


```python
# Retrait de la période dupliquée dans la première feuille

df_2009_2010_unique = df_2009_2010[
    df_2009_2010["InvoiceDate"] < "2010-12-01"
].copy()

# Consolidation des deux périodes

df_consolidated = pd.concat(
    [
        df_2009_2010_unique,
        df_2010_2011
    ],
    ignore_index=True
)

print("Dataset brut :", len(df_raw))
print("Dataset consolidé :", len(df_consolidated))
print("Lignes retirées :", len(df_raw) - len(df_consolidated))
```

```python
# verification de la periode : 
print("Date minimum :", df_consolidated["InvoiceDate"].min())
print("Date maximum :", df_consolidated["InvoiceDate"].max())
```

```python
# Recalcul des quantités negative :
negative_quantity = df_consolidated[
    df_consolidated["Quantity"] < 0
]

print(
    "Nombre de lignes avec quantité négative :",
    len(negative_quantity)
)

print(
    "Nombre de factures concernées :",
    negative_quantity["Invoice"].nunique()
)
```

```python
total_negative = len(negative_quantity)

negative_with_c = (
    negative_quantity["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

negative_without_c = total_negative - negative_with_c

print("Quantités négatives :", total_negative)
print("Avec facture C :", negative_with_c)
print("Sans facture C :", negative_without_c)

print(
    "Part avec facture C :",
    round(negative_with_c / total_negative * 100, 2),
    "%"
)
```

### constat sur les quantités négatives

Après suppression du chevauchement entre les deux feuilles:
le dataset consolidé contient 22 557 lignes avec une quantité négative, 
réparties sur 11 684 factures.

Parmi ces lignes :

- 19 164, soit 84,96 %, appartiennent à une facture dont l'identifiant commence par `C` ;
- 3 393, soit 15,04 %, possèdent une quantité négative sans facture commençant par `C`.

Les quantités négatives ne peuvent donc pas être assimilées automatiquement à des annulations clients.

Les 3 393 cas restants doivent être analysés avant de définir une règle de nettoyage.

### Conclusion sur les quantités négatives

L'analyse met en évidence deux types principaux de mouvements négatifs.

Sur les 22 557 lignes présentant une quantité négative :

- 19 164 lignes (84,96 %) appartiennent à une facture commençant par `C` ;

Les factures commençant par `C` seront considérées comme des annulations ou retours commerciaux.

### Décision :

- les mouvements négatifs seront conservés dans les données transformées afin de préserver l'information source

- les quantités négatives hors facture `C` seront identifiées comme mouvements négatifs non commerciaux ou non déterminés 
- ces mouvements ne seront pas intégrés aux indicateurs de ventes sans traitement spécifique.

### Cohérence des Client ID au niveau des factures

Aucune facture ne contient simultanément des lignes avec `Customer ID` renseigné et des lignes sans `Customer ID`.

L'absence d'identifiant client est donc cohérente au niveau de la facture entière.

**Décision :**

- aucun identifiant client ne sera imputé artificiellement 
- les factures sans `Customer ID` seront conservées pour les analyses commerciales lorsqu'elles correspondent à des transactions valides 
- elles seront exclues des analyses nécessitant une identification individuelle du client 


```python
# Nombre de lignes sans Customer ID par pays

display(
    missing_customer["Country"]
    .value_counts()
)
```

```python
# Taux de Customer ID manquants par pays

missing_by_country = (
    missing_customer["Country"]
    .value_counts()
    .rename("missing_customer_rows")
    .to_frame()
)

all_by_country = (
    df_consolidated["Country"]
    .value_counts()
    .rename("total_rows")
    .to_frame()
)

missing_by_country = missing_by_country.join(all_by_country)

missing_by_country["missing_rate_pct"] = (
    missing_by_country["missing_customer_rows"]
    / missing_by_country["total_rows"]
    * 100
)

display(
    missing_by_country
    .sort_values("missing_customer_rows", ascending=False)
)
```

Le phénomène est très fortement concentré au Royaume-Uni, avec 232 320 lignes concernées, soit 24,20 % des lignes britanniques.

## Synthèse de l'audit qualité

L'audit initial du dataset a permis d'identifier les principaux problèmes de qualité avant la construction du pipeline de données.

### 1. Chevauchement des sources

Les deux feuilles Excel contiennent une période commune du 1er au 9 décembre 2010.

22 523 lignes sont strictement identiques dans les deux feuilles.

**Règle :** une seule occurrence de cette période sera conservée.

### 2. Quantités négatives

22 557 lignes présentent une quantité négative après consolidation.

84,96 % appartiennent à une facture commençant par `C`, correspondant aux annulations.

Les autres lignes correspondent principalement à des mouvements tels que casse, perte, correction ou ajustement.

**Règle :** conserver les mouvements mais les classifier selon leur nature.

### 3. Prix négatifs

5 lignes présentent un prix négatif et correspondent à `Adjust bad debt`.

**Règle :** conserver mais identifier comme écritures non commerciales.

### 4. Prix nuls

6 024 lignes présentent un prix nul.

Une grande partie correspond à des corrections, mouvements internes ou observations non valorisées.

**Règle :** conserver mais ne pas considérer automatiquement comme ventes commerciales.

### 5. Prix extrêmes

La distribution des prix est fortement asymétrique et plusieurs prix élevés correspondent à des frais, commissions ou écritures manuelles.

**Règle :** aucun seuil arbitraire de prix ne sera utilisé pour supprimer les observations.

### 6. Customer ID manquant

235 287 lignes, soit 22,52 % du dataset, ne possèdent pas de `Customer ID`.

Aucune facture ne mélange lignes identifiées et non identifiées.

**Règle :**
- conserver ces transactions pour certaines analyses commerciales ;
- les exclure des analyses individuelles clients ;
- ne jamais imputer artificiellement un identifiant.

### 7. Description manquante

4 275 lignes, soit 0,41 %, ne possèdent pas de description.

Toutes ont un prix nul et un Customer ID manquant.

**Règle :** conserver pour la traçabilité mais exclure des analyses produits commerciales.

### 8. Doublons exacts

11 812 répétitions exactes subsistent après consolidation, soit 1,13 % du dataset.

**Règle :**
- conserver dans Bronze ;
- dédupliquer dans Silver en conservant une occurrence.
