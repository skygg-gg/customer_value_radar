```python
# On conserve la provenance de chaque ligne avant de réunir les deux feuilles.
# On ajoute SourcePeriod pour savoir de quelle feuille originale vient une ligne pour la retraçer.

df_2009_2010["SourcePeriod"] = "2009-2010"
df_2010_2011["SourcePeriod"] = "2010-2011"

df_raw = pd.concat(
    [df_2009_2010, df_2010_2011],
    ignore_index=True
)

print("Dimensions du dataset complet :", df_raw.shape)
```

```python
# Verification des doublons : 

duplicate_count = df_raw.duplicated().sum()

print("Nombre de doublons exacts :", duplicate_count)
```

```python
duplicate_rows = df_raw[df_raw.duplicated(keep=False)]

display(duplicate_rows.head(20))
```

```python
business_columns = [
    "Invoice",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "Price",
    "Customer ID",
    "Country"
]

duplicates_business = df_raw.duplicated(
    subset=business_columns,
    keep=False
)

print(
    "Lignes potentiellement dupliquées sur les colonnes métier :",
    duplicates_business.sum()
)
```

```python
display(
    df_raw.loc[duplicates_business]
    .sort_values(["InvoiceDate", "Invoice", "StockCode"])
    .head(30)
)
```

```python
# Verification chevauchement dates décembre 2010

overlap_2009_2010 = df_2009_2010[
    df_2009_2010["InvoiceDate"].between(
        "2010-12-01",
        "2010-12-09 23:59:59"
    )
]

overlap_2010_2011 = df_2010_2011[
    df_2010_2011["InvoiceDate"].between(
        "2010-12-01",
        "2010-12-09 23:59:59"
    )
]

print("Lignes décembre 2010 dans feuille 2009-2010 :", len(overlap_2009_2010))
print("Lignes décembre 2010 dans feuille 2010-2011 :", len(overlap_2010_2011))
```

```python
print("Factures feuille 2009-2010 :", overlap_2009_2010["Invoice"].nunique())

print("Factures feuille 2010-2011 :",overlap_2010_2011["Invoice"].nunique())
```

```python
common_invoices = set(overlap_2009_2010["Invoice"]).intersection(
    set(overlap_2010_2011["Invoice"])
)

print("Factures présentes dans les deux feuilles :", len(common_invoices))
```

Sur la periode les factures apparaissent sur les deux feuilles.

```python
# Colonnes à comparer entre les deux feuilles

compare_columns = [
    "Invoice",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "Price",
    "Customer ID",
    "Country"
]

# On trie les deux périodes 
overlap_a = (
    overlap_2009_2010[compare_columns]
    .sort_values(compare_columns, na_position="last")
    .reset_index(drop=True)
)

overlap_b = (
    overlap_2010_2011[compare_columns]
    .sort_values(compare_columns, na_position="last")
    .reset_index(drop=True)
)

print("Même nombre de lignes :", len(overlap_a) == len(overlap_b))
print("Contenu strictement identique :", overlap_a.equals(overlap_b))
```

### Constat de qualité : 
- les deux feuilles couvrent une periode commune du 1er au 9 décembre 2010
- Cette période contient 22 523 lignes et 1 088 factures dans chacune des deux feuilles. 
- Les 1 088 numéros de facture sont présents dans les deux sources.

Vérification pour voir si les lignes sont strictement identiques

```python
overlap_a.equals(overlap_b)
```

### Conclusion 

La période du 1er au 9 décembre 2010 est présente dans les deux feuilles du fichier original.

Les deux périodes contiennent exactement :

- 22 523 lignes ;
- 1 088 factures distinctes.

Après tri sur l’ensemble des colonnes métier, les deux sous-ensembles sont strictement identiques.
