```python
# On conserve la provenance de chaque ligne avant de réunir les deux feuilles.
# On ajoute SourcePeriod pour savoir de quelle feuille originale vient une ligne pour la retraçer.

df_2009_2010["SourcePeriod"] = "2009-2010"
df_2010_2011["SourcePeriod"] = "2010-2011"

df_raw = pd.concat(
    [df_2009_2010, df_2010_2011],
    ignore_index=True
)

print("Dimensions du dataset complet :", df_raw.shape)
```

```python
# Verification des doublons : 

duplicate_count = df_raw.duplicated().sum()

print("Nombre de doublons exacts :", duplicate_count)
```

```python
duplicate_rows = df_raw[df_raw.duplicated(keep=False)]

display(duplicate_rows.head(20))
```

```python
business_columns = [
    "Invoice",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "Price",
    "Customer ID",
    "Country"
]

duplicates_business = df_raw.duplicated(
    subset=business_columns,
    keep=False
)

print(
    "Lignes potentiellement dupliquées sur les colonnes métier :",
    duplicates_business.sum()
)
```

```python
# Verification chevauchement dates décembre 2010

overlap_2009_2010 = df_2009_2010[
    df_2009_2010["InvoiceDate"].between(
        "2010-12-01",
        "2010-12-09 23:59:59"
    )
]

overlap_2010_2011 = df_2010_2011[
    df_2010_2011["InvoiceDate"].between(
        "2010-12-01",
        "2010-12-09 23:59:59"
    )
]

print("Lignes décembre 2010 dans feuille 2009-2010 :", len(overlap_2009_2010))
print("Lignes décembre 2010 dans feuille 2010-2011 :", len(overlap_2010_2011))
```

```python
print("Factures feuille 2009-2010 :", overlap_2009_2010["Invoice"].nunique())

print("Factures feuille 2010-2011 :",overlap_2010_2011["Invoice"].nunique())
```

```python
common_invoices = set(overlap_2009_2010["Invoice"]).intersection(
    set(overlap_2010_2011["Invoice"])
)

print("Factures présentes dans les deux feuilles :", len(common_invoices))
```

Sur la periode les factures apparaissent sur les deux feuilles.

```python
# Colonnes à comparer entre les deux feuilles

compare_columns = [
    "Invoice",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "Price",
    "Customer ID",
    "Country"
]

# On trie les deux périodes 
overlap_a = (
    overlap_2009_2010[compare_columns]
    .sort_values(compare_columns, na_position="last")
    .reset_index(drop=True)
)

overlap_b = (
    overlap_2010_2011[compare_columns]
    .sort_values(compare_columns, na_position="last")
    .reset_index(drop=True)
)

print("Même nombre de lignes :", len(overlap_a) == len(overlap_b))
print("Contenu strictement identique :", overlap_a.equals(overlap_b))
```

### Constat de qualité : 
- les deux feuilles couvrent une periode commune du 1er au 9 décembre 2010
- Cette période contient 22 523 lignes et 1 088 factures dans chacune des deux feuilles. 
- Les 1 088 numéros de facture sont présents dans les deux sources.

Vérification pour voir si les lignes sont strictement identiques

```python
overlap_a.equals(overlap_b)
```

### Conclusion 

La période du 1er au 9 décembre 2010 est présente dans les deux feuilles du fichier original.

Les deux périodes contiennent exactement :

- 22 523 lignes ;
- 1 088 factures distinctes.

Après tri sur l’ensemble des colonnes métier, les deux sous-ensembles sont strictement identiques.


lors de l’intégration des deux feuilles, une seule occurrence de cette période sera conservée afin d’éviter un double comptage des transactions.

La feuille Year 2010-2011 sera conservée comme source de référence pour cette période de chevauchement.

### Comprendre les quantités negatives : 

```python
# Analyse des quantités négatives

negative_quantity = df_raw[df_raw["Quantity"] < 0]

print("Nombre de lignes avec quantité négative :", len(negative_quantity))
print("Nombre de factures concernées :", negative_quantity["Invoice"].nunique())
```

```python
display(
    negative_quantity[
        [
            "Invoice",
            "StockCode",
            "Description",
            "Quantity",
            "Price",
            "Customer ID",
            "Country",
            "InvoiceDate"
        ]
    ].head(30)
)
```

```python
# Verification des factures : 

negative_invoice_check = (
    negative_quantity["Invoice"]
    .astype(str)
    .str.startswith("C")
    .value_counts()
)

display(negative_invoice_check)
```

```python
# Calculons les proportions : 

total_negative = len(negative_quantity)

negative_with_c = (
    negative_quantity["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

negative_without_c = total_negative - negative_with_c

print("Quantités négatives :", total_negative)
print("Avec facture C :", negative_with_c)
print("Sans facture C :", negative_without_c)

print(
    "Part avec facture C :",
    round(negative_with_c / total_negative * 100, 2),
    "%"
)
```

## Construction du dataset consolidé

On a chevauchement de 22 523 lignes entre les deux feuilles pour la période du 1er au 9 décembre 2010.

Afin d'éviter tout doublons, une seule occurrence de cette période est conservée.

Le dataset consolidé contient :
- la feuille 2009-2010 jusqu'au 30 novembre 2010 ;
- l'intégralité de la feuille 2010-2011.


```python
# Retrait de la période dupliquée dans la première feuille

df_2009_2010_unique = df_2009_2010[
    df_2009_2010["InvoiceDate"] < "2010-12-01"
].copy()

# Consolidation des deux périodes

df_consolidated = pd.concat(
    [
        df_2009_2010_unique,
        df_2010_2011
    ],
    ignore_index=True
)

print("Dataset brut :", len(df_raw))
print("Dataset consolidé :", len(df_consolidated))
print("Lignes retirées :", len(df_raw) - len(df_consolidated))
```

```python
# verification de la periode : 
print("Date minimum :", df_consolidated["InvoiceDate"].min())
print("Date maximum :", df_consolidated["InvoiceDate"].max())
```

```python
# Recalcul des quantités negative :
negative_quantity = df_consolidated[
    df_consolidated["Quantity"] < 0
]

print(
    "Nombre de lignes avec quantité négative :",
    len(negative_quantity)
)

print(
    "Nombre de factures concernées :",
    negative_quantity["Invoice"].nunique()
)
```

```python
total_negative = len(negative_quantity)

negative_with_c = (
    negative_quantity["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

negative_without_c = total_negative - negative_with_c

print("Quantités négatives :", total_negative)
print("Avec facture C :", negative_with_c)
print("Sans facture C :", negative_without_c)

print(
    "Part avec facture C :",
    round(negative_with_c / total_negative * 100, 2),
    "%"
)
```

### constat sur les quantités négatives

Après suppression du chevauchement entre les deux feuilles:
le dataset consolidé contient 22 557 lignes avec une quantité négative, 
réparties sur 11 684 factures.

Parmi ces lignes :

- 19 164, soit 84,96 %, appartiennent à une facture dont l'identifiant commence par `C` ;
- 3 393, soit 15,04 %, possèdent une quantité négative sans facture commençant par `C`.

Les quantités négatives ne peuvent donc pas être assimilées automatiquement à des annulations clients.

Les 3 393 cas restants doivent être analysés avant de définir une règle de nettoyage.

```python
# On isole les cas sans le C : 
negative_without_c_df = negative_quantity[
    ~negative_quantity["Invoice"]
    .astype(str)
    .str.startswith("C")
].copy()

print(
    "Nombre de lignes avec quantité négative sans facture C :",
    len(negative_without_c_df)
)
```
